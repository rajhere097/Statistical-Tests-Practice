**Every Hypothesis test is build on two statements**

```Null Hypothesis - It takes a default or initial assumption Its says no effect, no change.```
```Alternative Hypothesis - The statement we will take or challenge. There is an effect or change.```

Example 1: Medicine Effect

Null: The new medicine has no effect on blood pressure.

Alternative: The new medicine does reduce blood pressure.

**Types of Error**

**Type 1 Error = You think something is there, but actually its not. Example: You say a patient has cancer, but really they dont.**

**Type 2 Error = You say nothing was there, but actually there was something(you missed it). Example: You say a patient doesn’t have cancer, but really they do.**

## Example 1 ##

Example: Fire Alarm System

Null Hypothesis (H₀): There is no fire.

Alternative Hypothesis (H₁): There is a fire.

Scenario 1 

The alarm rings, but in reality there is no fire.
Is this Type I or Type II?

Ans: Type 1 error because there was no fire but we assumped there was a fire. 


Scenario 2 

The alarm does not ring, but in reality there is a fire.
Is this Type I or Type II?

Ans: Type2 error as there was really a fire but we said no fire, so we missed it, critical. 

**What is P Value?**

p-value = how much chance the null hypothesis is right.

If p is small (p ≤ 0.05) → Null is very unlikely → Reject H₀ → Accept alternative.

If p is big (p > 0.05) → Null is still possible → Keep H₀ → Do not accept alternative.

## Example 1

Testing if average sales = ₹10,000.

p = 0.02 → very small → Null must go → Sales are not ₹10,000.

p = 0.40 → big → Keep Null → No strong evidence that sales differ from ₹10,000.


# Example 2

Example: Patient Cancer Test

H₀ (Null): Patient does not have cancer.

H₁ (Alternative): Patient has cancer.

Case 1:
The test gives a p-value = 0.01 (very small, less than 0.05).

Means: Null hypothesis is unlikely, accepts alternative.

Decision: Reject H₀ → Conclude the patient likely has cancer.

Case 2:
The test gives a p-value = 0.40 (large, greater than 0.05).

Means: There’s still a good chance that the null (no cancer) is true.

Decision: Fail to reject H₀ → Conclude the patient likely does not have cancer (or not enough evidence).


**T test - One Sample**

````python
## T test - One Sample

import pandas as pd\

df = pd.read_csv(r'C:\Users\LENOVO\Downloads\Walmartsales Dataset.csv')
df.shape
````

````python
df.isnull().sum()
````

````python
df['Total']
````

````python
Sample = np.random.choice(df['Total'], size=30, replace=True)
print(Sample)
````

## lets assume boss has given the sample and he asks please check if an average sale is Rs1000 or not. If real life scenario no one will give you a data
## of 1 million and ask you to check They will only give you fraction of it to run the test


````python
from scipy import stats

#boss assumption
mu = 1000

## null hypothesis = Rs 1000 is true
## alternative hypothesis = Rs 1000 is not true
t_stats, p_value = stats.ttest_1samp(Sample, mu)

t_stats = float(t_stats)
p_value = float(p_value)

print(f't_stats is {t_stats}')
print(f'p_value is {p_value}')
````

## P value is 0.0 so if p value is <=0.05 null hypothesis is false so we can say our boss average sale is not Rs1000
